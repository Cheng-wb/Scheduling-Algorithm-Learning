**将所有功能用函数封装**

````python
def get_jobs_num(jobs):
    return len(jobs)

def get_total_processing_time(jobs):
    return sum(job['processing_time'] for job in jobs)

def get_shortest_job(jobs):
    return min(jobs, key = lambda job: job['processing_time'])

def get_longest_job(jobs):
    return max(jobs, key = lambda job: job['processing_time'])

def get_average_processing_time(jobs):
    return get_total_processing_time(jobs) / get_jobs_num(jobs)

def sort_jobs_ascending(jobs):
    return sorted(jobs, key = lambda job: job['processing_time'])

def sort_jobs_descending(jobs):
    return sorted(jobs, key = lambda job: job['processing_time'], reverse=True)

def main():
    jobs = [
        {"id": "J1", "processing_time": 5},
        {"id": "J2", "processing_time": 2},
        {"id": "J3", "processing_time": 8},
        {"id": "J4", "processing_time": 3},
        {"id": "J5", "processing_time": 6},
        ]
    print(f"任务数量：{get_jobs_num(jobs)}")
    print(f"加工总时间：{get_total_processing_time(jobs)}")
    print(f"平均加工时间：{get_average_processing_time(jobs):.2f}")

    shortest = get_shortest_job(jobs)
    longest = get_longest_job(jobs)
    print(f"最短任务：{shortest['id']} {shortest['processing_time']}")
    print(f"最长任务：{longest['id']} {longest['processing_time']}")

    ascending = sort_jobs_ascending(jobs)
    descending = sort_jobs_descending(jobs)
    print(f"时间升序：{[job['id'] for job in ascending]}")
    print(f"时间降序：{[job['id'] for job in descending]}")

main()
````

#### 定义 Machine

````python
from dataclasses import dataclass

@dataclass
class Machine:
    machine_id: str
    available_time: int = 0    # 什么时候可以执行下一个任务

    def schedule_job(self, job: Job):
        start_time = max(self.available_time, job.release_time)
        finish_time = start_time + job.processing_time
        self.available_time = finish_time
        return start_time, finish_time

Machine1 = Machine('M1')
job1 = Job('J1', 5, release_time=2)
start, finish = Machine1.schedule_job(job1)
print(f"Job {job1.job_id} starts at {start} and finishes at {finish}. Machine available at {Machine1.available_time}.")
````

#### 理解 Job 和 Operation 的区别

真实生产中，生产一个零件 J1，需要：
> 切割 → 钻孔 → 打磨  

那么：
```
Job J1
│
├── Operation 1：切割
├── Operation 2：钻孔
└── Operation 3：打磨
```
所以：
> Job 是完整任务，Operation 是 Job 内部的一道工序。  

例如：
> J1:  
> O11 → O12 → O13  

如果：
> O11 → M1 → 3  
> O12 → M2 → 5  
> O13 → M3 → 2  

代表：
| Operation | Machine | Processing Time |
| :-------: | :-----: | :-------------: |
| O11       | M1      |               3 |
| O12       | M2      |               5 |
| O13       | M3      |               2 |

且必须按照： O11 &rarr; O12 &rarr; O13  顺序完成。

#### 理解三种调度问题

1. Single Machine Scheduling  
只有一台机器：M1  
   ``` 
   J1 ───┐  
   J2 ───┼──→ 排序 → M1  
   J3 ───┤  
   J4 ───┘  
   ```

   核心问题：Job 按照什么顺序执行？  
   - FCFS (First-Come, First-Served)  
     - J1 → J2 → J3  
   - SPT(Shortest Process Time)/ SJF(Shortest Job First)  
     - J2 → J1 → J3  

2. Parallel Machine Scheduling  
现在有2台机器：M1, M2  都能处理 Job  
   ``` 
           ┌──→ M1  
    Jobs ──┤  
           └──→ M2  
   ```
   
     - 需要决定两件事
       1. Job 分给哪台机器？  
       2. 每台机器内部按照什么顺序执行？  

   例如：
   ``` 
   M1: J1 → J3  
   M2: J2 → J4  
   ```

3. Job Shop Scheduling  
Job 有多个 Operation:
   ```
   J1:  
   M1(3) → M2(5) → M3(2)  
   J2:  
   M2(4) → M1(2) → M3(6)  
   ```
- 两条重要约束：  
    - 工序顺序约束（Precedence Constraint）：J1 必须 O11 → O12 → O13  
    - 机器资源约束（Resource Constraint）：O11 用 M1，O22 也用 M1，不能同时执行  



#### 理解 Release Time 和 Due Time

- 完成一个通用的评价函数

#### 时间概念

对于：
```
J1

release_time = 2
processing_time = 5
due_date = 10
```
如果： start_time = 3  
则：Completion Time = Start Time + Processing Time = 8  
- 时间轴：
    ```
    0    2    3                  8       10
    |----|----|------------------|--------|
         ↑    ↑                  ↑        ↑
      release start           finish    due
    ```
- 区分：
    ```
    release_time
    任务什么时候可以进入加工

    start_time
    任务实际什么时候开始

    processing_time
    任务需要加工多久

    completion_time
    任务什么时候完成

    due_date
    任务希望什么时候完成
    ```
- 对应变量：
    ```
    r_j = release time
    S_j = start time
    p_j = processing time
    C_j = completion time
    d_j = due date
    ```


#### Completion Time

单机调度,有：
```
J1 processing_time = 5
J2 processing_time = 2
J3 processing_time = 4
```
调度顺序：
```
J1 → J2 → J3
```
时间轴：
```
0        5   7       11
|--------|---|--------|
   J1      J2    J3
```
所以：
```
C1 = 5
C2 = 7
C3 = 11
```
平均完成时间：
$$\hat C = \frac{5+7+11}{3}=7.67$$

Completion Time 越小，意味着任务越早完成。

但是单个任务完成得早，不代表整体调度一定更好，所以我们需要更多指标


#### Flow Time

一个任务从进入系统，到最终完成，一共经历了多久。

$$ F_j = C_j - r_j $$
如果：
```
release_time = 2
completion_time = 8
```
那么： Flow Time = 8 - 2 = 6  

特别地，如果所有任务： release_time = 0  
则：$$ F_j = C_j $$


#### Makespan

从调度开始，到最后一个任务完成所需要的总时间。
$$ C_{max} = \max_j C_j $$

#### 理解 Lateness 和 Tardiness

- Lateness（延迟）：它表示实际完工时间相对预期期限的带符号偏差（可正、可负、可为零）。
$$ L_j = C_j - d_j $$
- Tardiness(延期)：它表示实际完工比期限晚了多少（只算正数，提前或准时完成则记为 0）。
$$ T_j = \max(0, C_j - d_j) $$

如果：  
 > J1    
 > completion_time = 12  
 > due_date = 10  

则：
> Lateness = 12 - 10 = 2  
> Tardiness = 2  

如果：
> completion_time = 8  
> due_date = 10  

则：
> Lateness = 8 - 10 = -2  
> Tardiness = 0


#### 平均延期和最大延期

如果：
```
J1 tardiness = 0
J2 tardiness = 3
J3 tardiness = 2
```

- 平均延期：关注整体平均延期水平
$$ \hat T = \frac{0+3+2}{3} = 1.67 $$
- 最大延期：关注最严重的延期任务
$$ T_{max} = \max_j T_j $$

#### Machine Utilization

机器利用率表示：
> 机器真正工作的时间，占整个调度周期的比例。
$$ U = \frac{\text{Machine Busy Time}}{\text{Makespan}} $$

#### 把 Job 扩展成“调度结果对象”

增加：
> start_time: int = 0  
> completion_time: int = 0  

增加方法：  
> def flow_time(self)  
> def lateness(self)  
> def tardiness(self)  


````python
from dataclasses import dataclass


@dataclass
class Job:
    job_id: str
    processing_time: int

    release_time: int = 0
    due_date: int = 0

    start_time: int = 0
    completion_time: int = 0

    @property
    def flow_time(self):
        return (
            self.completion_time
            - self.release_time
        )

    @property
    def lateness(self):
        return (
            self.completion_time
            - self.due_date
        )

    @property
    def tardiness(self):
        return max(
            0,
            self.lateness
        )
    
````

这样 Job 本身就可以知道：
```
processing_time
release_time
due_date
start_time
completion_time
flow_time
lateness
tardiness
```
